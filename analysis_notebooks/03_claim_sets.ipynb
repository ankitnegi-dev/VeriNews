# 03 - Claim sets

Counts and dates for the three labeled claim files. No network is used.

```python
import pandas as pd
from pathlib import Path
base = Path("..")
claims = pd.read_csv(base / "tests/data/claims.csv")
fresh = pd.read_csv(base / "tests/data/fresh_claims_20.csv")
third = pd.read_csv(base / "tests/data/third_claims_20.csv")
print("claims.csv shape:", claims.shape)
print("fresh_claims_20.csv shape:", fresh.shape)
print("third_claims_20.csv shape:", third.shape)
```

## Label balance

```python
for name, df in (("claims", claims), ("fresh", fresh), ("third", third)):
    print(name, "labels:")
    print(df["label"].value_counts().to_string())
    print()
```

## Category counts

```python
for name, df in (("claims", claims), ("fresh", fresh), ("third", third)):
    print(name, "categories:")
    print(df["category"].value_counts().to_string())
    print()
```

## Dev and test split (claims.csv)

```python
print(claims["split"].value_counts().to_string())
print()
print("by split and label:")
print(claims.groupby(["split", "label"]).size().to_string())
```

## Event dates

```python
for name, df in (("claims", claims), ("fresh", fresh), ("third", third)):
    dates = pd.to_datetime(df["event_date"], errors="coerce", dayfirst=True)
    print(name, "date range:", dates.min(), "to", dates.max())
    print(dates.dt.to_period("M").value_counts().sort_index().to_string())
    print()
```

## Note on traps

The fresh and third sets were built to include retrieval traps: adjacent matches
in the same cricket series, fixture pages instead of match reports, and claims
whose wording differs from headlines.
